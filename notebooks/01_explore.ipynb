# 03 Minutes per player-season

```python
import matplotlib.pyplot as plt

# One row per player-season, 1980+ NBA: keep the xTM row (listed first), drop the per-team rows
seasons = per100[(per100["lg"] == "NBA") & (per100["season"] >= 1980)]
seasons = seasons.drop_duplicates(["season", "player_id"], keep="first")

CUTOFF = 500
under = (seasons["mp"] < CUTOFF).sum()
print(f"{len(seasons):,} player-seasons; {under:,} ({under / len(seasons):.0%}) under {CUTOFF} minutes")

fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(seasons["mp"], bins=range(0, 3700, 100), color="#2a78d6", edgecolor="white", linewidth=1)
ax.axvline(CUTOFF, color="#52514e", linestyle="--", linewidth=1.5)
ax.annotate(f"{CUTOFF} min cutoff\n{under / len(seasons):.0%} of player-seasons below",
            xy=(CUTOFF, ax.get_ylim()[1] * 0.92), xytext=(10, 0), textcoords="offset points",
            va="top", color="#52514e", fontsize=9)
ax.set_title("Minutes per player-season, NBA 1980\u20132026", loc="left", fontsize=11)
ax.set_xlabel("Minutes played in the season")
ax.set_ylabel("Player-seasons")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", color="#e6e5e1", linewidth=0.8)
ax.set_axisbelow(True)
plt.show()
```

# 04 Nulls, 1980–2026

```python
# Same rows as the minutes chart (1980+ NBA, one row per player-season), with Advanced columns
# and the raw attempt counts from Player Totals joined on
def player_seasons(df):
    df = df[(df["lg"] == "NBA") & (df["season"] >= 1980)]
    return df.drop_duplicates(["season", "player_id"], keep="first")

KEYS = ["season", "player_id", "team"]
adv_cols = [c for c in advanced.columns if c not in per100.columns]
ps = (player_seasons(per100)
      .merge(player_seasons(advanced)[KEYS + adv_cols], on=KEYS)
      .merge(player_seasons(playerTotal)[KEYS + ["fga", "x3pa", "x2pa", "fta", "tov", "ast", "orb"]], on=KEYS))

no_shots = (ps["fga"] == 0) & (ps["fta"] == 0)
reasons = {
    "gs": ("only some teams tracked it in 1980–81", ps["season"] <= 1981),
    "x3p_percent": ("0 three-point attempts", ps["x3pa"] == 0),
    "x2p_percent": ("0 two-point attempts", ps["x2pa"] == 0),
    "fg_percent": ("0 field-goal attempts", ps["fga"] == 0),
    "e_fg_percent": ("0 field-goal attempts", ps["fga"] == 0),
    "x3p_ar": ("0 field-goal attempts", ps["fga"] == 0),
    "f_tr": ("0 field-goal attempts", ps["fga"] == 0),
    "ft_percent": ("0 free-throw attempts", ps["fta"] == 0),
    "ts_percent": ("0 FGA and 0 FTA", no_shots),
    "tov_percent": ("0 FGA, FTA and turnovers", no_shots & (ps["tov"] == 0)),
    "o_rtg": ("0 FGA, FTA, TOV, AST and ORB", ps[["fga", "fta", "tov", "ast", "orb"]].sum(axis=1) == 0),
    "pos": ("missing in source", ps["pos"].isna()),
}

rows = []
for col in ps.columns[ps.isna().any()]:
    null = ps[col].isna()
    reason, cond = reasons.get(col, ("0 minutes played", ps["mp"] == 0))
    rows.append({
        "column": col,
        "nulls": null.sum(),
        "reason": reason,
        "explained": (null & cond).sum(),       # nulls where the reason holds
        "reason_not_null": (~null & cond).sum(), # reason holds but value present (0 = exact match)
        "max_mp": ps.loc[null, "mp"].max(),
        "qualified_nulls": (null & (ps["mp"] >= 500)).sum(),
    })
pd.DataFrame(rows).set_index("column").sort_values("nulls", ascending=False)
```
